```python
df = pd.read_csv("../data/application_train.csv")
print("shape:", df.shape)
print("Dataset loaded successfully")
```

```python
df.head()
```

```python
df.info()
```

```python
print(df["TARGET"].value_counts())
print("\nDefault rate:", round(df['TARGET'].mean() * 100, 2), "%")
```

```python
# missing values
missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({
    'missing_count': missing,
    'missing_percent': round(missing_percent, 2)
})
missing_df = missing_df[missing_df['missing_count'] > 0].sort_values('missing_percent', ascending=False)
print("Columns with missing values:", len(missing_df))
print(missing_df.head(20))
```

```python
plt.figure(figsize=(12, 8))
missing_df[missing_df['missing_percent'] > 40]['missing_percent'].plot(kind='bar')
plt.title('Columns with more than 40% missing values')
plt.xlabel('Columns')
plt.ylabel('Missing %')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
cat_cols = df.select_dtypes(include='object').columns
print("Categorical columns:", len(cat_cols))
print("\n")
for col in cat_cols:
    print(f"{col}: {df[col].nunique()} unique values")
    print(df[col].value_counts().head(3))
    print("---")
```

```python
df.describe()
```

```python
print(df['DAYS_EMPLOYED'].describe())
print("\nPositive values count:", (df['DAYS_EMPLOYED'] > 0).sum())
```

```python
plt.figure(figsize=(12,8))
corr = df.select_dtypes(include=[np.number]).corr()['TARGET'].sort_values(ascending=False)
print("Top 10 features correlated with TARGET:")
print(corr.head(10))
print("\nBottom 10 features correlated with TARGET:")
print(corr.tail(10))
```

```python
plt.figure(figsize=(6,4))
df['TARGET'].value_counts().plot(kind='bar', color=['green','red'])
plt.title('Loan Default Distribution')
plt.xlabel('Target (0=No Default, 1=Default)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
